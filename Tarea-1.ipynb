### Descripción de la base de datos

#### La base de datos muestra los resultados del monitoreo satelital de sargazo flotante en el Caribe mexicano en el año 2025, realizado por el Sistema de Información y Análisis Marino Costero (SIMAR+) de la Comisión Nacional para el Conocimiento y Uso de la Biodiversidad (CONABIO).
#### Está compuesta por 365 registros y 8 variables asociadas. Las cuatro primeras corresponden a información temporal: fecha, año, mes y día de observación. Las tres siguientes proveen información respecto a  la estimación de biomasa húmeda por día en toneladas, la combinación de la estimación de biomasa húmeda de dos días y la superficie asociada a la estimación por ambos días en km2. Finalmente, la octava variable indica la región geográfica del monitoreo. De acuerdo con la exploración realizada, todas las columnas contienen valores de observación, es decir, no hay valores nulos.
